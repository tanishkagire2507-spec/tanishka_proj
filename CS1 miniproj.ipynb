```python
import sqlite3

conn = sqlite3.connect("shop.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS products (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    price INTEGER
)
""")

cursor.execute("DELETE FROM products")  
cursor.execute("INSERT INTO products (name, price) VALUES ('Laptop', 50000)")
cursor.execute("INSERT INTO products (name, price) VALUES ('Phone', 20000)")
cursor.execute("INSERT INTO products (name, price) VALUES ('Headphones', 3000)")

conn.commit()
conn.close()

print("Database ready!")
```

```python
def search_product_vulnerable(user_input):
    conn = sqlite3.connect("shop.db")
    cursor = conn.cursor()
    
    
    query = f"SELECT * FROM products WHERE name = '{user_input}'"
    print("Executing Query:", query)
    
    cursor.execute(query)
    results = cursor.fetchall()
    
    conn.close()
    return results
```

```python
search_product_vulnerable("Laptop")
```

```python
search_product_vulnerable("' OR 1=1 --")
```

```python
def search_product_safe(user_input):
    conn = sqlite3.connect("shop.db")
    cursor = conn.cursor()
    
    query = "SELECT * FROM products WHERE name = ?"
    print("Executing Safe Query:", query)
    
    cursor.execute(query, (user_input,))
    results = cursor.fetchall()
    
    conn.close()
    return results
```

```python
search_product_safe("' OR 1=1 --")

```
